# Retail Sales Forecasting

## Objective
Analyze historical retail sales data and forecast future sales trends using
a moving average time-series technique.

## Tools Used
- Python
- Pandas
- Matplotlib


```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("retail_sales_data.csv")
df['Month'] = pd.to_datetime(df['Month'])
df.head()
```

```python
df['Moving_Average'] = df['Sales'].rolling(window=3).mean()
df
```

```python
plt.figure(figsize=(8,4))
plt.plot(df['Month'], df['Sales'], label="Actual Sales")
plt.plot(df['Month'], df['Moving_Average'], label="Forecast (Moving Avg)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Retail Sales Forecasting")
plt.legend()
plt.show()
```

## Key Insights
- Sales show a steady upward trend over time.
- Moving average smooths short-term fluctuations.
- Forecast suggests continued growth in future sales.
- This analysis can help businesses plan inventory and demand.
