# Customer Segmentation Using Clustering

## Objective
Segment customers based on age, income, and spending behavior using K-Means
clustering to support data-driven marketing strategies.

## Tools Used
- Python
- Pandas
- Scikit-learn
- Matplotlib


```python
import os
os.environ["OMP_NUM_THREADS"] = "1"
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("customer_data.csv")
df.head()
```

```python
X = df[['Age', 'Annual_Income', 'Spending_Score']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)
df
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['Annual_Income'], df['Spending_Score'], c=df['Cluster'])
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")
plt.show()
```

## Key Insights
- Customers are grouped into three distinct segments.
- High-spending customers can be targeted for premium offers.
- Low-engagement customers may require retention strategies.
- Customer segmentation enables personalized marketing decisions.
