# Social Media Sentiment Analysis

## Objective
Analyze customer reviews to classify sentiment as positive, negative, or neutral
using natural language processing techniques.


```python
%pip install textblob
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob
```

```python
df = pd.read_csv(
    "reviews_data.csv",
    engine="python"
)
df
```

```python
df['Review'] = df['Review'].astype(str)
df['Review']
```

```python
from textblob import TextBlob

def get_sentiment(text):
    try:
        polarity = TextBlob(text).sentiment.polarity
        if polarity > 0:
            return "Positive"
        elif polarity < 0:
            return "Negative"
        else:
            return "Neutral"
    except:
        return "Neutral"

df['Sentiment'] = df['Review'].apply(get_sentiment)
df


```

```python
sentiment_counts = df['Sentiment'].value_counts()
sentiment_counts
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sentiment_counts.plot(kind='bar')
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.show()
```

## Key Insights
- Customer reviews can be categorized into positive and negative sentiments.
- Majority sentiment indicates overall customer satisfaction.
- Sentiment analysis helps organizations understand public opinion efficiently.
