# Student Performance Analysis

## Objective
Analyze students' academic performance to calculate subject-wise averages,
identify top-performing students, validate data quality, and visualize results.

## Tools Used
- Python
- Pandas
- Matplotlib
- Seaborn


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

plt.style.use('default')
```

```python
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
math_avg = df['math score'].mean()
reading_avg = df['reading score'].mean()
writing_avg = df['writing score'].mean()

print("Average Math Score:", math_avg)
print("Average Reading Score:", reading_avg)
print("Average Writing Score:", writing_avg)
```

```python
df['total_score'] = df['math score'] + df['reading score'] + df['writing score']

top_students = df.sort_values(by='total_score', ascending=False).head(5)
top_students
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Check for negative marks
negative_values = (df[['math score', 'reading score', 'writing score']] < 0).sum()
negative_values
```

```python
subjects = ['Math', 'Reading', 'Writing']
averages = [math_avg, reading_avg, writing_avg]

plt.figure(figsize=(8,5))
sns.barplot(x=subjects, y=averages)
plt.title("Average Scores by Subject")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.show()
```

## Key Insights
- Students show different performance levels across subjects.
- Overall performance can be evaluated using combined subject scores.
- Data validation confirms that the dataset is clean and reliable.
- Visual analysis helps in comparing subject-wise performance easily.
