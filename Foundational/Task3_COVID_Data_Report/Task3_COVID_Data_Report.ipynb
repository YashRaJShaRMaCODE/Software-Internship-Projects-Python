# COVID-19 Data Analysis Report

## Objective
The objective of this project is to analyze COVID-19 data at a global level
to understand daily trends in confirmed cases, deaths, and recoveries over time.

## Tools Used
- Python
- Pandas
- Matplotlib
- Seaborn


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('default')
```

```python
df = pd.read_csv("covid_19_clean_complete.csv")
df.head()

```

```python
# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Check for missing values
df.isnull().sum()
```

```python
# Aggregate global data by date
global_data = df.groupby('Date')[['Confirmed', 'Deaths', 'Recovered']].sum().reset_index()

global_data.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(global_data['Date'], global_data['Confirmed'], label='Confirmed Cases')
plt.plot(global_data['Date'], global_data['Deaths'], label='Deaths')
plt.plot(global_data['Date'], global_data['Recovered'], label='Recovered')

plt.title("Global COVID-19 Daily Trends")
plt.xlabel("Date")
plt.ylabel("Number of Cases")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Display last few days to observe trend direction
global_data.tail()
```

## Key Insights
- Global confirmed cases increased rapidly during peak pandemic periods.
- Death and recovery trends followed similar time-based patterns.
- Time-series visualization helps understand pandemic progression clearly.
- Such analysis can support public health planning and response strategies.
