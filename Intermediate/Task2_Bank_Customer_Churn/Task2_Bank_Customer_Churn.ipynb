# Bank Customer Churn Analysis

## Objective
Analyze bank customer data to identify factors influencing customer churn
using exploratory data analysis and visualization.

## Tools Used
- Python
- Pandas
- Matplotlib
- Seaborn


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('default')
```

```python
df = pd.read_csv("Churn.csv")
df.head()
```

```python
df.info()
```

```python
df['Exited'].value_counts()
```

```python
sns.countplot(x='Exited', data=df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=False, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Exited', y='Age', data=df)
plt.title("Age vs Customer Churn")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Exited', y='Balance', data=df)
plt.title("Balance vs Customer Churn")
plt.show()
```

## Key Insights
- Churned customers are generally older than retained customers.
- Higher account balance shows correlation with churn behavior.
- Heatmap reveals relationships between churn and key customer attributes.
- These insights can help banks improve customer retention strategies.
